```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("placement.csv")
df.head()
```

```python
plt.figure(figsize=(14,4))
plt.subplot(121)
sns.distplot(df['cgpa'],kde=True)
plt.title('Age PDF')

plt.subplot(122)
sns.distplot(df['placement_exam_marks'],kde=True)
plt.title('Age QQ plot')
plt.show()
```
